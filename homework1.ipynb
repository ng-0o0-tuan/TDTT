```python
### bai 5
ch = str(input("nhap chu cai: "))
def dich_chuyen(ch):
    if ch.isupper():
        ch = ch.lower()
        moc = ord('a')
    else:
        ch = ch.upper()
        moc = ord('A')
    vi_tri_cu = ord(ch) - moc
    vi_tri_moi = vi_tri_cu - 1
    return chr(vi_tri_moi + moc)
print(dich_chuyen(ch))
```

```python
### bai 6
import math
def tinh_toan():
    a, b, c = map(float, input("độ dài 3 cạnh: ").split())
    if  a < b+c and b < a+c and c < a+b:
        p = (a+b+c)/2
        return math.sqrt(p*(p-a)*(p-b)*(p-c))
    else:
        print("khong phai tam giac")
print(tinh_toan())
```
